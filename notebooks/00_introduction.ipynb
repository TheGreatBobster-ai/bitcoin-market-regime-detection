# Abstract

Bitcoin is characterized by repeating bull and bear markets, periods of strong speculation, rapid growth, crashes and changing market conditions. Identifying such market regimes can help to better understand Bitcoin market behaviour and potentially improve investment and risk management decisions. 

The goal of this project is to investigate whether meaningful Bitcoin market regimes can be identified using Unsupervised Machine Learning techniques. For this purpose, a dataset containing market, network and on-chain variables was collected and transformed using three different feature engineering approaches. Principal Component Analysis (PCA) was applied to reduce dimensionality before performing clustering.

Three clustering algorithms were evaluated: K-Means, Gaussian Mixture Models (GMM) and HDBSCAN. The resulting clusters were compared using clustering metrics, visual inspection and economic reasoning.

The results show that Bitcoin market regimes can be identified successfully using Unsupervised Machine Learning. One of the most important findings is that feature engineering had, with some exceptions, a larger impact on the final results than the clustering algorithm itself. While all algorithms were applied to the same underlying information, the selected feature representation significantly influenced the resulting regime structures. 

Dataset C based on rolling Z-Scores combined with K-Means clustering produced the most convincing results (K-Means C-k4). The identified clusters consistently captured bull markets, bear markets and transitional phases reliably while also showing surprisingly early signals of upcoming market regime changes. GMM produced similar but generally less reliable results, while HDBSCAN struggled to identify meaningful regime structures.

# 0. Introduction

Bitcoin and other cryptocurrencies exhibit particularly strong structural market phases. Examples include bull and bear markets, accumulation and distribution periods, liquidity events and speculative peaks. While the internet is full of attempts to identify such phases using price charts and technical indicators, these approaches often ignore the large amount of additional information that public blockchains provide transparently.

The core idea of this project is therefore to move beyond price-based regime identification and instead use a broader set of market, network and on-chain variables combined with Unsupervised Machine Learning techniques to uncover hidden market structures. If such structures can be identified reliably, they may not only help explain past market behaviour but potentially also acting as early signals of upcoming regime changes and trend reversals.

The research question of this project is therefore:

*Can Unsupervised Machine Learning methods identify useful Bitcoin market regimes using market, network and on-chain data, that can potentially also act as early warning signals of upcoming change?*

To answer this question, an initial dataset was obtained containing market variables (price, returns, volume), short- and longterm-demand on-chain metrics (active addresses, hash rate, fees, transaction activity), exchange-related metrics (exchange inflows, outflows and balances) as well as valuation measures such as MVRV. In addition, features were engineered to incorporate historical developments and longer-term market dynamics - the step that took up by far the most time, while also yielding the highest return - followed by scaling and dimensionality reduction using PCA. Finally, three different Unsupervised Machine Learning Algortihms were appliead and evaluated: K-Means, Gaussian Mixture Models and HDBSCAN. 

Based on Bitcoin's recurring market cycles, strong speculation and herding behaviour, I expected the algorithms to identify multiple distinct market regimes rather than a simple bull-versus-bear classification. Furthermore, I initially expected HDBSCAN to perform best, as different market regimes could potentially form distinct density regions within the feature space. Regarding feature engineering, I expected Dataset B to perform best, as the longer growth horizon appeared to capture regime-like structures more effectively during the exploratory analysis.

The results showed a different picture. While multiple meaningful market regimes could indeed be identified successfully, Dataset C based on rolling Z-Scores consistently produced the strongest results. In addition, K-Means clearly outperformed the other clustering methods, while HDBSCAN struggled to identify useful regime structures. Not only did K-Means find reliable and consistent market regimes, but also possible early regime change signals up to months before the trend reversal happened.


The project follows the following workflow:

1. Explore and preprocess Bitcoin market, network and on-chain data.
2. Construct three different feature datasets using growth rates, moving averages and rolling Z-Scores.
3. Standardize the features and apply Principal Component Analysis (PCA).
4. Apply and compare K-Means, Gaussian Mixture Models (GMM) and HDBSCAN.
5. Evaluate the resulting clusters using quantitative metrics, visual inspection and economic reasoning.
6. Select and discuss the preferred model.
